# Learn error handling, writing code that anticipates something might go wrong, and decides what to do about it, instead of letting the program crash.

Up to Day 8, any mistake or unexpected situation, a missing dictionary key, dividing by zero, the wrong type of data, would stop your code completely with an error message. Day 9 introduces try/except: you tell Python "attempt this, but if a specific problem happens, do this instead," so your program can keep running smoothly even when something unexpected occurs.

This matters for bioinformatics because real datasets are rarely clean. A gene you're looking for might be missing from a specific dataset, a file might not exist where you expect it, a value might come in as text when you expected a number. Error handling lets your code deal with these situations gracefully, instead of crashing the entire analysis over one bad data point.

# A basic try/except

```python
gene_expression = {"APP": 4.2, "MAPT": 3.8, "PSEN1": 2.1, "APOE": 5.6}

try:
  print(gene_expression["TREM2"])
except KeyError:
    print("That gene isn't in the dictionary")
```

# Wrap the safe lookup in a reusable function

```python
def safe_lookup(gene, expression_dict):
    try:
        return expression_dict[gene]
    except KeyError:
        return None


result = safe_lookup("TREM2", gene_expression)
if result is None:
   print("Gene not found")
else:
    print(f"Expression value: {result}")

```

# Handling more than one type of error

```python
def divide_expression(value1, value2):
    try:
      return value1/value2
    except ZeroDivisionError:
        return"Cannot divide by zero"
    except TypeError:
        return "Both values must be numbers"

print(divide_expression(4.2, 0))
print(divide_expression(4.2, 2))
```

# Practice: using safe_lookup on multiple genes, including missing ones

```python
genes_to_check = ["APP", "TREM2", "MAPT", "GRN"]
for gene in genes_to_check:
    result = safe_lookup(gene, gene_expression)
    if result is None:
        print(f"{gene}: not found in dataset" )
    else:
       print(f"{gene}: {result}")

```
