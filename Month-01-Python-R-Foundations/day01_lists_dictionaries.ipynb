# Day 1: Python Basics for Bioinformatics

## Topic
Introduction to Python Lists

### Objective
Learn how to create a list, access its elements, and determine its length using Python.

### Example
We will use three genes associated with Alzheimer's disease:
- APP
- MAPT
- PSEN1

### Learning Outcomes
By the end of today's lesson, I should be able to:
- Create a Python list.
- Access elements using indexing.
- Count the number of elements using `len()`.
- Display output using `print()`.
- add a new gene using append(0)
- learn how to add a new entry and update exisiting entry
- find a gene with expression above  certain threshold


```python
# A list of gene names
genes = ["APP", "MAPT", "PSEN1"]

print(genes)
print(genes[0])       # access first item
print(len(genes))     # how many items
```

```python
genes.append("APOE")  # add a new gene
print(genes)
```

```python
for gene in genes:
    print(gene)
```

```python
# A dictionary mapping gene name -> expression level
gene_expression = {
    "APP": 4.2,
    "MAPT": 3.8,
    "PSEN1": 2.1
}

print(gene_expression)
print(gene_expression["MAPT"])   # access value by key
```

```python
gene_expression["APOE"] = 5.6   # add a new entry
gene_expression["APP"] = 4.5    # update an existing one
print(gene_expression)
```

```python
for gene, expression in gene_expression.items():
    print(f"{gene}: {expression}")
```

```python
# Find genes with expression above a threshold
threshold = 3.0
high_expression_genes = [gene for gene, exp in gene_expression.items() if exp > threshold]

print("Genes with high expression:", high_expression_genes)
```
