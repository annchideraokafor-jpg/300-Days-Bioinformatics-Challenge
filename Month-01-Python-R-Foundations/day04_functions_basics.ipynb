# Learn how to write functions — reusable blocks of code you define once and call whenever you need them, instead of rewriting the same logic repeatedly.

Day 3 introduced if/elif/else to make decisions, but that decision logic (checking if expression is high, moderate, or low) had to be rewritten inside every loop. Day 4 wraps that logic into a function, a named, reusable piece of code you can call on any value, as many times as you want, without retyping it.

This matters for bioinformatics because real analysis code constantly reuses the same operations, on different genes, different samples, different datasets. Functions let you write one trustworthy piece of logic and apply it everywhere, instead of copy-pasting (and risking mistakes) each time.

# Turn a repeated task into a function

```python
def get_status(expression):
   if expression > 4.5:
     return "very high"
   elif expression > 3.0:
    return "moderate"
   else:
      return "low"

print(get_status(4.2))
print(get_status(2.1))
```

# Use the function on all your genes (instead of rewriting the if/elif each time)

```python
gene_expression = {"APP": 4.2, "MAPT": 3.8,"PSEN1":2.1, "APOE":5.6 }

for gene, exp in gene_expression.items():
    status = get_status(exp)
    print(f"{gene}: {exp} - {status}")
```

# A function that takes more than one input

```python
def compare_expression(gene1, gene2, expression_dict):
    val1 = expression_dict[gene1]
    val2 = expression_dict[gene2]
    if val1 > val2:
       return f"{gene1} is more highly expressed than {gene2}"
    elif val2 > val1:
       return f"{gene2} is more highly expressed than {gene1}"
    else:
        return f"{gene1} and {gene2} have equal expression"

print(compare_expression("APP", "MAPT", gene_expression))

```

# A function that returns a list (reusable version of Day 1's filtering)

```python
def filter_high_expression(expression_dict, threshold=3.5):
  result=[]
  for gene, exp in expression_dict.items():
    if exp > threshold:
      result.append(gene)
  return result

print(filter_high_expression(gene_expression))
print(filter_high_expression(gene_expression, threshold=4.0))
```
