# Day 13 — Practice Problem: Gene Report Generator

The problem: Given a messy list of gene data (as raw text strings), clean it up, classify each gene, handle any errors gracefully, and generate a formatted report, all in one script.

```python
# Day 13: Combining Week 2 skills into one practice problem

# Messy raw data (as if read from a file, Day 6 style)

raw_data = ["   app:4.2", "MAPT:3.8", "psen1:2.1", "apoe:5.6  ", "bace1:not_a_number"]

# Step 1: Clean and parse each line ( from Day 11 — string methods)
def parse_gene_line(line):
    try:
       gene, value = line.strip().split(":")
       gene = gene.strip().upper()
       value = float(value.strip())
       return gene, value
    except ValueError:
       return None, None

# Step 2: Classify expression ( from Day 3/4 — conditionals + functions)
def get_status(expression):
    if expression > 4.5:
       return "very high"
    elif expression >3.0:
       return "moderate"
    else:
       return "low"

# Step 3: Process everything, handling bad data gracefully ( from Day 9 — error handling)
gene_expression = {}
for line in raw_data:
    gene, value = parse_gene_line(line)
    if gene is None:
        print(f"Skipped invalid entry: '{line.strip()}'")
    else:
        gene_expression[gene] = value

print("\nCleaned data:", gene_expression)

# Step 4: Build a report using a list comprehension ( from Day 10)

report = [f"{gene}: {value} - {get_status(value)}" for gene, value in gene_expression.items()]

print("\nFinal Report:")
for line in report:
    print(line)
```
