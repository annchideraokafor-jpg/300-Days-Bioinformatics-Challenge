# Day 8: Review — reinforcing Week 1 concepts with a new dataset

```python
gene_expression = {
    "APP": 4.2, "MAPT": 3.8, "PSEN1": 2.1, "APOE": 5.6,
    "BACE1": 3.1, "TREM2": 2.9
}

def get_status(expression):
    if expression > 4.5:
       return "very high"
    elif expression > 3.0:
       return  "moderate"
    else:
        return "low"



# Practice: loop + function + conditional together

for gene, exp in gene_expression.items():
    print(f"{gene}: {exp} - {get_status(exp)}")


# Practice: sorting with a new, larger dataset
sorted_genes = sorted(gene_expression.items(), key=lambda x: x[1], reverse=True)
print("\nSorted by expression:")
print(sorted_genes)

# Practice: filtering with the reusable function pattern

def filter_high_expression(expression_dict, threshold=3.5):
    return [gene for gene, exp in expression_dict.items() if exp > threshold]

print("\nHighly expressed genes:", filter_high_expression(gene_expression))

# Practice: writing results to a file
with open("day8_review.txt", "w") as file:
   for gene, exp in sorted_genes:
      file.write(f"{gene}: {exp} - {get_status(exp)}\n")

print("\nReview saved to file.")


```
