# Learn list comprehensions, a shorter, more compact way to build a list from a loop, all in a single line, instead of writing a full multi-line loop with .append().

# The "old" way: a regular loop

```python
gene_expression = {"APP": 4.2, "MAPT": 3.8, "PSEN1": 2.1, "APOE": 5.6, "BACE1": 3.1}

high_expression = []
for gene, exp in gene_expression.items():
    if exp > 3.5:
        high_expression.append(gene)

print(high_expression)
```

# The same thing as a list comprehension

```python
high_expression_comp = [gene for gene, exp in gene_expression.items() if exp > 3.5]
print(high_expression_comp)
```

# List comprehension without a condition

```python
all_values_doubled = [exp *2 for exp in gene_expression.values()]
print(all_values_doubled)
```

# List comprehension building formatted strings

```python
gene_summary = [f"{gene}: {exp}" for gene, exp in gene_expression.items()]
print(gene_summary)
```
