# Building on yesterday's lists and dictionaries, today I'm using them together to actually do something with the data, sorting genes by expression level, and pulling basic stats (max, min, average). This is a small step toward what real RNA-seq analysis looks like: not just storing data, but asking questions of it.

#Combine a list and dictionary into something more useful

```python
genes = ["APP", "MAPT", "PSEN1", "APOE"]
gene_expression = {"APP": 4.2, "MAPT": 3.8, "PSEN1": 2.1, "APOE": 5.6}

# Build a new list showing gene + expression together
gene_summary = []
for gene in genes:
    gene_summary.append(f"{gene}: {gene_expression[gene]}")

print(gene_summary)
```

# Sorting

```python
# Sort genes by expression level, highest first
sorted_genes = sorted(gene_expression.items(), key=lambda x: x[1], reverse=True)
print(sorted_genes)
```

# Basic statistics on the value

```python
expression_values = list(gene_expression.values())

print("Highest expression:", max(expression_values))
print("Lowest expression:", min(expression_values))
print("Average expression:", sum(expression_values) / len(expression_values))
```

# A tiny "report" using string formatting

```python
for gene, exp in sorted_genes:
    print(f"{gene} has an expression level of {exp}")
```
