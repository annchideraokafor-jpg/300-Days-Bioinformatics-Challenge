# Day 12 — Intro to Classes & Objects

Get a first, gentle exposure to Object-Oriented Programming (classes and objects), without needing to master it yet.

So far, you've stored gene data in dictionaries (a name mapped to a value) and written separate functions to work on that data (like get_status()). A class is a different approach: it bundles related data and the functions that work on that data into a single blueprint. From that blueprint, you create individual objects, each one representing a single gene, carrying both its own information (name, expression value) and its own behavior (like classifying its own status) together in one place.

# Defining a simple class

```python
class Gene:
   def __init__(self, name, expression):
      self.name = name
      self.expression = expression


# Creating an object (an "instance") from the class

app_gene = Gene("APP", 4.2)

print(app_gene.name)
print(app_gene.expression)
```

# Adding a method (a function that belongs to the class)

```python
class Gene:
   def __init__(self, name, expression):
       self.name = name
       self.expression = expression

   def get_status(self):
       if self.expression > 4.5:
           return "very high"
       elif self.expression > 3.0:
           return "moderate"
       else:
           return "low"

app_gene = Gene("APP", 4.2)
print(app_gene.get_status())
```

# Creating multiple objects from the same class

```python
mapt_gene = Gene("MAPT", 3.8)
psen1_gene = Gene("PSEN1", 2.1)

genes = [app_gene, mapt_gene, psen1_gene]

for gene in genes:
    print(f"{gene.name}: {gene.expression} - {gene.get_status()}")
```

# Comparing the class approach to the dictionary approach

```python
# The dictionary way (what you've been doing all week)
gene_expression = {"APP": 4.2,"MAPT":3.8, "PSEN1": 2.1}

# The class way (today's new approach)
gene_objects = [Gene("APP", 4.2), Gene("MAPT",3.8), Gene("PSEN1", 2.1)]

print("Dictionary approach:", gene_expression)
print("Class approach:", [(g.name, g.expression) for g in gene_objects])
```
