# Learn how to make code respond differently depending on the data, using if / elif / else statements (conditionals).

Days 1-2 focused on storing data (lists, dictionaries) and processing it the same way for every item (looping, sorting, basic stats). Day 3 introduces decision-making: code that checks a condition and does one thing if it's true, another if it's false.

This matters for bioinformatics because almost every real analysis involves filtering or categorizing data based on a threshold, for example, keeping only genes above a certain expression level, flagging samples that fail quality control, or classifying results as significant vs. not significant based on a cutoff. Today's exercise (sorting genes into "high," "moderate," or "low" expression) is a simplified version of this same filtering logic, and it's structurally the same pattern used later in real differential expression analysis.

# A basic if/else

```python
gene_expression = {"APP": 4.2, "MAPT": 3.8, "PSEN1": 2.1, "APOE": 5.6}

expression = gene_expression["MAPT"]

if expression > 3.5:
    print("MAPT is highly expressed")
else:
    print("MAPT is not highly expressed")
```

# Looping through the dictionary, checking each gene

```python
for gene, exp in gene_expression.items():
    if exp > 3.5:
        print(f"{gene} is highly expressed ({exp})")
    else:
        print(f"{gene} is low expressed ({exp})")
```

# Adding a third condition (elif)

```python
for gene, exp in gene_expression.items():
    if exp > 4.5:
        status = "very high"
    elif exp > 3.0:
        status = "moderate"
    else:
        status = "low"
    print(f"{gene}: {exp} — {status}")
```

# Combining the loop + condition into a filtered list (ties back to Day 1's example)

```python
high_expression_genes = []

for gene, exp in gene_expression.items():
    if exp > 3.5:
        high_expression_genes.append(gene)

print("Highly expressed genes:", high_expression_genes)
```
