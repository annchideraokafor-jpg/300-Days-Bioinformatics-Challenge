Day 11 Goal

Learn how to clean, split, join, and search text data using Python's built-in string methods.

Real biological data files are rarely perfectly formatted, gene names might have inconsistent capitalization, extra spaces, or need to be split apart from a combined line (like a CSV row: "APP,MAPT,PSEN1"). Today's exercise practices exactly that kind of cleanup, taking messy, inconsistent text and standardizing it, a step that almost always happens before any real analysis can begin.

# Basic string operations

```python
gene_name = "  app  "

print(gene_name.upper())      # convert to uppercase
print(gene_name.lower())      # convert to lowercase
print(gene_name.strip())      # remove extra whitespace
print(len(gene_name))         # count characters (including spaces)


```

# Cleaning messy gene names (a real bioinformatics problem)

```python
messy_genes = ["APP", "mapt", "PSEN1", "apoe"]

cleaned_genes = []
for gene in messy_genes:
    cleaned = gene.strip().upper()
    cleaned_genes.append(cleaned)

print(cleaned_genes)
```

# Splitting and joining strings

```python
gene_line = "APP, MAPT, PSEN1, APOE"

# Split a single string into a list
gene_list = gene_line.split(" , ")
print(gene_list)

# Join a list back into a single string

rejoined = " , ".join(gene_list)
print(rejoined)
```

# Checking string content (useful for filtering/searching)

```python
gene_description = "APP is associated with amyloid plaque formation"

print("amyloid" in gene_description)      # checks if a substring exists
print(gene_description.startswith("APP"))  # checks the beginning
print(gene_description.replace("APP", "the APP gene")) # replaces text
```
