# Day 6 Goal

Learn File I/O (Input/Output) — how to write data to a file, and read data back from a file, in Python.

Days 1–5 all worked with data typed directly into the code (the genes list, the gene_expression dictionary). Day 6 introduces the missing piece: how to save that data outside your code, into an actual file, and how to load data back in from a file someone else (or your past self) created.



# Writing data to a file

```python
gene_expression = {"APP": 4.2,"MAPT": 3.8,"PSEN1": 2.1,"APOE": 5.6}

with open("gene_expression.txt", "w") as file:
    for gene, exp in gene_expression.items():
        file.write(f"{gene}: {exp}\n")



print("File written succesfully")
```

# Reading the file back

```python
with open("gene_expression.txt", "r") as file:
    contents = file.read()

print(contents)
```

# Reading a file line by line (useful for larger files later)

```python
with open("gene_expression.txt", "r") as file:
   for line in file:
       print(line.strip())    # .strip() removes the extra newline character
```

# Reading the file back into a dictionary (turning text back into usable data)

```python
loaded_expression = {}

with open("gene_expression.txt", "r") as file:
    for line in file:
        gene, value = line.strip() .split(": ")
        loaded_expression[gene] = float(value)


print(loaded_expression)

```
