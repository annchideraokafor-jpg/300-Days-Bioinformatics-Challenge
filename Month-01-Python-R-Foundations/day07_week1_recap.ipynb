# Day 7 — Week 1 Recap

Goal: Combine everything from Days 1-6 (lists, dictionaries, sorting, stats, conditionals, functions, refactoring, file I/O) into one small, complete mini-program.

```python
# Day 7: Week 1 Recap — combining everything learned this week

gene_expression = {"APP": 4.2,"MAPT": 3.8,"PSEN1": 2.1,"APOE": 5.6}

 # Function to classify expression level (Day 3 + 4)

def get_status(expression):
     if expression > 4.5:
         return "very high"
     elif expression > 3:
         return "moderate"
     else:
         return "low"


  # Loop through, classify, and build a report (Day 2 + 5)
report_lines = []
for gene, exp in gene_expression.items():
       status = get_status(exp)
       report_lines.append(f"{gene}: {exp} - {status}")


  # Sort by expression, highest first (Day 2)
sorted_genes = sorted(gene_expression.items(), key=lambda x: x[1], reverse=True)

  # Basic stats (Day 2)

values = list(gene_expression.values())
average = sum(values) / len(values)

  # Write the report to a file (Day 6)

with open("week1_report.txt", "w") as file:
      file.write("week 1 gene expression report\n")
      file.write("=============================\n")
      for line in report_lines:
          file.write(line + "\n")
      file.write(f"\nHighest expression: {sorted_genes[0]}\n")
      file.write(f"Average expression: {average:.2f}\n")

 # Read it back to confirm it saved correctly (Day 6)
with open("week1_report.txt", "r") as file:
    print(file.read())

```
