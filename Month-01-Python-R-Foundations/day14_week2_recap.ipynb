```python
# Day 14: Week 2 Recap — combining error handling, comprehensions, strings, and classes

class Gene:
    def __init__(self, name, expression):
        self.name = name
        self.expression = expression

    def get_status(self):
        if self.expression > 4.5:
          return "very high"
        elif self.expression > 3.0:
          return "moderate"
        else:
          return "low"

raw_data = ["app:4.2", "mapt:3.8", "psen1:2.1","apoe:5,6","bad_entry"]

gene_objects = []
for line in raw_data:
    try:
        name, value = line.split(":")
        gene_objects.append(Gene(name.strip().upper(), float(value.strip())))
    except ValueError:
        print(f"Skipped: '{line}'")


# List comprehension to build the final report

report = [f"{g.name}: {g.expression} - {g.get_status()}" for gin gene_objects]

for line in report:
    print(line)


```
