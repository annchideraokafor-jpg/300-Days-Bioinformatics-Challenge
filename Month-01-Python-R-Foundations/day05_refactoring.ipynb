# Learn to refactor: rewrite existing code to be cleaner and more reusable, while keeping the same result.

Day 4 introduced functions, but even after learning them, it's easy to still repeat yourself (like checking each gene's status one at a time, copy-pasting the same logic). Day 5 shows the "before and after": what code looks like without using your function properly (repetitive, copy-pasted) versus with it (short, reusable, looped).

This matters because in real analysis work, code is rarely written perfectly the first time. Refactoring, going back and improving structure once something works, is a normal and important part of writing good code, especially as projects get bigger and you need to trust that your logic is consistent everywhere it's applied.

# Before: repeated, unrefactored code (what it might have looked like without functions)

```python
gene_expression = {"APP": 4.2, "MAPT": 3.8, "PSEN1":2.1, "APOE": 5.6}

# Check APP Manually
if gene_expression["APP"] > 4.5:
    print("APP: very high")
elif gene_expression["APP"] > 3.0:
    print("APP: moderate")
else:
    print("APP: low")

 # Checking MAPT manually
if gene_expression["MAPT"] > 4.5:
     print("MAPT: very high")
elif gene_expression["MAPT"] > 3.0:
     print("MAPT: moderate")
else:
     print("MAPT: low")
```

# After: refactored using your Day 4 function


```python
def get_status(expression):
    if expression > 4.5:
       return "very high"
    elif expression > 3.0:
       return "moderate"
    else:
       return "low"
for gene, exp in gene_expression.items():
    print(f"{gene} {get_status(exp)}")
```
