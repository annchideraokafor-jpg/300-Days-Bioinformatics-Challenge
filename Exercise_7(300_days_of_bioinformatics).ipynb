# Classes + String Cleaning

```python
class Gene:
    def __init__(self,name,expression):
       self.name = name
       self.expression = expression

    def clean_name(self):
        return self.name.strip().upper()

gene1 = Gene("  app ", 4.2)
gene2 = Gene("mapt", 3.8)
gene3 = Gene(" PSEN1", 2.1)

genes = [gene1, gene2, gene3]
for gene in genes:
    print(f"{gene.clean_name()}: {gene.expression}")
```
